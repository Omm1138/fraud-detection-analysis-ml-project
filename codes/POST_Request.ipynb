```python

# This script sends a POST request to the Flask server to make a prediction
# Replace the "data" dictionary with the input values you want to predict on

import requests

url = "http://127.0.0.1:8000/predict/"
data = {"features": [.01, 0.05, 0.50, 0.005]}  # Replace with real input values

try:
	response = requests.post(url, json=data)
	response.raise_for_status()  # Raise an HTTPError for bad responses (4xx and 5xx)
	print(response.json())
except requests.exceptions.RequestException as e:
	print(f"An error occurred: {e}")
```
